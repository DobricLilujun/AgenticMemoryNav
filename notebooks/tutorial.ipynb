## Convert GLB to USD

This guide explains how to convert a `.glb` file (binary GLTF format) to `.usd` (Universal Scene Description) format using available tools.

```python
import subprocess
from pathlib import Path

PROJECT_ROOT = Path('/home/snt/projects/AgenticMemoryNav')
INTERN_DIR = PROJECT_ROOT / 'external-lib' / 'InternScenes'
BASE_DIR = INTERN_DIR / 'InternScenes'

glb_path = BASE_DIR / 'tutorial' / 'examples' / 'bbox_scenes' / 'bbox_scene.glb'
if not glb_path.exists():
    raise FileNotFoundError(f'Missing GLB: {glb_path}')

usd_path = BASE_DIR / 'tutorial' / 'examples' / 'bbox_scenes' / 'bbox_scene.usd'
usd_path.parent.mkdir(parents=True, exist_ok=True)

cmd = [
    '/home/snt/isaacsim/python.sh',
    str(INTERN_DIR / 'glb2usd_headless.py'),
    '--file', str(glb_path),
    '--out', str(usd_path),
]
print('Running:', ' '.join(cmd))
r = subprocess.run(cmd, capture_output=True, text=True)
print(r.stdout[-2000:])
if r.stderr:
    print('STDERR:\n', r.stderr[-2000:])
print(f'converted={usd_path.exists()} rc={r.returncode}')
```
