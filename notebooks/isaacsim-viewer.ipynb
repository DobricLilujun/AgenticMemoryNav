```python
# standalone_preview_usd.py
import sys
from isaacsim.simulation_app import SimulationApp

# 1) 先启动 SimulationApp（必须在其他 isaacsim 导入之前）
headless = False  # 有 GUI 就 False；纯命令行/无显示器用 True + 可能还需要 xvfb
kit = SimulationApp({"headless": headless})

# 2) 打开 USD 场景
usd_path = "/absolute/path/to/your_scene.usd"
import omni
omni.usd.get_context().open_stage(usd_path)

# 3) 可选：获取 stage 检查内容
stage = omni.usd.get_context().get_stage()
print("Stage loaded:", stage.GetRootLayer().identifier)

# 4) 运行主循环，让 GUI/渲染更新
while kit.is_running():
    kit.update()
```
